Task 1: Delivery Fee Array Calculator (NumPy)

- Build a NumPy-based program that computes and analyses delivery fees for a batch of food orders using vectorised operations — no Python loops.

- Generate a 1D NumPy array of 25 random delivery distances between 1.0 km and 15.0 km using np.random.uniform() with a fixed seed of 42.

```python
import numpy as np

np.random.seed(42)

distance = np.random.uniform(1.0, 15.0, 25)

print(distance)
```

- Compute the delivery fee for every order using the formula: fee = Rs 20 + Rs 5 × distance — applied as a single vectorised expression, not a loop.

```python
del_fee = (20+5)*distance
print(del_fee)
```

- Use boolean indexing to extract and print all orders where the delivery fee exceeds Rs 60, showing both the distance and the fee for each qualifying order.

```python
# Boolean indexing
condition = del_fee > 60

# Print distance and fee for qualifying orders
print("Orders with delivery fee greater than $60:")
# print("Distance:", distance[condition])
# print("Delivery Fee:", del_fee[condition])


print("Distance (km)   Fee (₹)")
print("----------------------")
print(np.column_stack((distance[condition], del_fee[condition])))
```

- Print the minimum, maximum, mean, and standard deviation of the complete fee array using NumPy statistical functions.

```python
print("Minimum delivery fee:", np.min(del_fee))
print("Maximum delivery fee:", np.max(del_fee))
print("Average delivery fee:", np.mean(del_fee))
print("Standard deviation of delivery fees:", np.std(del_fee))
```
