SCENARIO 3
You are developing a monthly performance report for a food delivery 
startup. The report must present three pieces of information side by side: (a) total orders placed per month over a year, 
(b) the proportion of orders by cuisine type 
(Fast Food, Indian, Chinese, Desserts), and 
(c) the distribution of delivery times across all orders.

# Which chart should be used for each metric?
- A line chart is the best choice because it shows the trend 
of total orders over the months.

# Why is that chart the best choice?
- A pie chart is the best choice because it shows the percentage 
  or proportion of each cuisine type.
- A histogram is the best choice because it shows how delivery 
  times are distributed.

# How do we arrange all three charts using plt.subplots()?
- Since there are 3 charts, place them in one row with three columns.

# Should we use a shared x-axis (sharex=True)?
- A shared x-axis is not appropriate because the three charts use 
  different types of data. 
- The line chart uses months, the pie chart has no x-axis, 
  and the histogram uses delivery times.

```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "Order_ID": [
        101,102,103,104,105,106,107,108,
        109,110,111,112,113,114,115,116,
        117,118,119,120,121,122,123,124
    ],

    "Month": [
        "Jan","Jan","Feb","Feb","Mar","Mar",
        "Apr","Apr","May","May","Jun","Jun",
        "Jul","Jul","Aug","Aug","Sep","Sep",
        "Oct","Oct","Nov","Nov","Dec","Dec"
    ],

    "Cuisine_Type": [
        "Fast Food","Indian","Chinese","Desserts",
        "Fast Food","Indian","Chinese","Desserts",
        "Fast Food","Indian","Chinese","Desserts",
        "Fast Food","Indian","Chinese","Desserts",
        "Fast Food","Indian","Chinese","Desserts",
        "Fast Food","Indian","Chinese","Desserts"
    ],

    "Delivery_Time_Mins": [
        25,30,22,18,
        28,35,26,20,
        24,32,27,19,
        23,34,29,21,
        26,31,28,22,
        25,33,30,20
    ]
}

df = pd.DataFrame(data)

print(df)
```

```python
# ----------------------------
# Create and Display Charts
# ----------------------------

month_order = ['Jan','Feb','Mar','Apr','May','Jun','Jul','Aug','Sep','Oct','Nov','Dec']
orders_per_month = df['Month'].value_counts().reindex(month_order, fill_value=0)
cuisine_counts = df['Cuisine_Type'].value_counts()
delivery_times = df['Delivery_Time_Mins']

# 1. Line Chart

orders_per_month = (
    df["Month"]
    .value_counts()
    .reindex([
        "Jan","Feb","Mar","Apr","May","Jun",
        "Jul","Aug","Sep","Oct","Nov","Dec"
    ])
)

plt.figure(figsize=(8,5))

plt.plot(
    orders_per_month.index,
    orders_per_month.values,
    marker="o"
)

plt.title("Orders Per Month")
plt.xlabel("Month")
plt.ylabel("Number of Orders")
plt.grid(True)

plt.show()



```

```python
# 2. Pie Chart
cuisine_counts = df["Cuisine_Type"].value_counts()

plt.figure(figsize=(7,7))

plt.pie(
    cuisine_counts.values,
    labels=cuisine_counts.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Cuisine Distribution")

plt.show()
```

```python
# 3. Histogram


delivery_times = df["Delivery_Time_Mins"]

plt.figure(figsize=(8,5))

plt.hist(
    delivery_times,
    bins=5,
    edgecolor="black"
)

plt.title("Delivery Time Distribution")
plt.xlabel("Delivery Time (mins)")
plt.ylabel("Frequency")

plt.show()
```
