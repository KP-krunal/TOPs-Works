# Section A
SCENARIO 5
- You are designing a visual analysis of customer ratings for a food delivery app. 
- Your dataset has 10,000 rows. Each row belongs to one of three restaurant categories (Fast Food, Fine Dining, Casual) and has a customer rating on a discrete 1–5 scale. 
- Your manager wants to understand how ratings are distributed within each category and whether any category produces more extreme outliers.

# Answer
- I would recommend sns.boxplot() because the dataset has only 3 categories and ratings are limited to a discrete 1–5 scale. 
- Boxplots clearly show the median, spread, and extreme outliers, which is exactly what the manager wants. With 10,000 rows, a boxplot also gives a clean summary of each category. 
- A violinplot would be preferable when the data is more continuous and we want to see the detailed distribution/density shape.
