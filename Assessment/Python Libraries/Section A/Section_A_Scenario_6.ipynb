# Section A
Scenario 6
- You are building a feature analysis report for a food delivery platform's data science team before a regression model is trained. 
- Your DataFrame contains eight numeric columns: order_value, distance_km, delivery_time_mins, rating, discount_applied, peak_hour_flag, restaurant_age_years, and driver_experience_years.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    "order_value": [500, 700, 450, 900, 650, 800, 300, 750],
    "distance_km": [5, 8, 3, 10, 6, 9, 2, 7],
    "delivery_time_mins": [25, 35, 20, 45, 30, 40, 15, 38],
    "rating": [4.5, 4.0, 4.8, 3.5, 4.2, 3.8, 4.9, 4.1],
    "discount_applied": [50, 70, 30, 100, 60, 80, 20, 75],
    "peak_hour_flag": [1, 1, 0, 1, 0, 1, 0, 1],
    "restaurant_age_years": [5, 8, 3, 10, 6, 9, 2, 7],
    "driver_experience_years": [2, 5, 1, 7, 4, 6, 1, 5]
}

df = pd.DataFrame(data)
df
```

```python
corr = df.corr()

sns.heatmap(corr, annot=True, cmap='coolwarm', fmt=".2f")

plt.title("Correlation Heatmap")
plt.show()
```
